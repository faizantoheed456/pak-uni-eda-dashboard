```python
def data_cleaning_pipeline(uni_data):
    """
    Advanced Data Cleaning Pipeline for Pakistan Universities Dataset.
    Logs all imputations via flag columns, resolves regional urban biases, 
    cleans categorical quirks, and optimizes memory formats.
    """
    # Create a deep copy to preserve the raw input data
    df_clean = uni_data.copy()
    
    # 1. Handle disguised missing values
    df_clean.replace(['N/A', 'na', 'unknown', ''], pd.NA, inplace=True)
    
    # List of all columns we expect to impute/fill
    impute_target_cols = [
        'HEC_Ranking', 'Industry_Tie_Ups', 'Fee_Per_Semester_PKR', 'Scholarship_Students', 
        'International_Students', 'Avg_CGPA', 'No_of_Buildings', 'Lab_Count', 
        'Dropout_Rate_Pct', 'Research_Papers_Published', 'BS_Students', 
        'PhD_Students', 'MS_MPhil_Students', 'Avg_Student_Background',
        'Entry_Test_Tier', 'Entry_Test_Passing_Criteria'
    ]
    
    # Create boolean flag columns tracking missingness before cleaning
    for col in impute_target_cols:
        if col in df_clean.columns:
            df_clean[f'{col}_is_imputed'] = df_clean[col].isna()
    
    # 2. Revert HEC_Ranking cleanly (Preserve real ranks, map missing/0 to 'None')
    df_clean['HEC_Ranking'] = df_clean['HEC_Ranking'].astype(str).str.replace(r'\.0$', '', regex=True).str.strip()
    df_clean['HEC_Ranking'] = pd.to_numeric(df_clean['HEC_Ranking'], errors='coerce').fillna(0).astype(int)
    df_clean['HEC_Ranking'] = df_clean['HEC_Ranking'].astype(str).replace({'0': 'None'})
    
    # 3. Handle Numerical Columns via University-Specific Medians (Fallback to Global Median)
    numeric_impute_cols = [
        'Industry_Tie_Ups', 'Fee_Per_Semester_PKR', 'Scholarship_Students', 
        'International_Students', 'Avg_CGPA', 'No_of_Buildings', 
        'Lab_Count', 'Dropout_Rate_Pct', 'Research_Papers_Published'
    ]
    df_clean[numeric_impute_cols] = df_clean.groupby('UNI_ID')[numeric_impute_cols].transform(lambda x: x.fillna(x.median()))
    df_clean[numeric_impute_cols] = df_clean[numeric_impute_cols].fillna(df_clean[numeric_impute_cols].median())
    
    # 4. Handle Student Level Relationships Logically (BS, PhD, MS/MPhil)
    calculated_bs = df_clean['Total_Enrollment'] - (df_clean['MS_MPhil_Students'].fillna(0) + df_clean['PhD_Students'].fillna(0))
    df_clean['BS_Students'] = df_clean['BS_Students'].fillna(calculated_bs)
    
    calculated_phd = df_clean['Total_Enrollment'] - (df_clean['BS_Students'].fillna(0) + df_clean['MS_MPhil_Students'].fillna(0))
    df_clean['PhD_Students'] = df_clean['PhD_Students'].fillna(calculated_phd)
    
    calculated_ms_mphil = df_clean['Total_Enrollment'] - (df_clean['BS_Students'].fillna(0) + df_clean['PhD_Students'].fillna(0))
    df_clean['MS_MPhil_Students'] = df_clean['MS_MPhil_Students'].fillna(calculated_ms_mphil)
    
    # 5. Handle Categorical Columns via Regional Mode (Mitigates Urban Bias)
    province_modes = df_clean.groupby('Province')['Avg_Student_Background'].transform(
        lambda x: x.fillna(x.mode()[0] if not x.mode().empty else df_clean['Avg_Student_Background'].mode()[0])
    )
    df_clean['Avg_Student_Background'] = df_clean['Avg_Student_Background'].fillna(province_modes)
    
    # Clean entry test details based on structural filter requirements
    test_mask = df_clean['Entry_Test_Required'].str.strip().str.lower() != 'no'
    for col in ['Entry_Test_Tier', 'Entry_Test_Passing_Criteria']:
        mode_val = df_clean.loc[test_mask, col].mode()[0]
        df_clean.loc[test_mask & df_clean[col].isna(), col] = mode_val
        df_clean.loc[~test_mask & df_clean[col].isna(), col] = 'None'
        
    # 6. Clean Text Whitespaces
    string_cols = df_clean.select_dtypes(include=['object']).columns
    for col in string_cols:
        df_clean[col] = df_clean[col].str.strip()
        
    # 7. Optimize Numeric Types (Downcasting Memory)
    for col in df_clean.select_dtypes(include=['int64', 'float64']).columns:
        df_clean[col] = pd.to_numeric(df_clean[col], downcast='integer' if df_clean[col].dtype == 'int64' else 'float')
        
    # 8. Convert Text Columns to Categorical Data Type
    categorical_cols = [
        'Type', 'Province', 'City', 'HEC_Category', 'HEC_Ranking', 'Department_Name', 
        'Department_Category', 'Avg_Student_Background', 'Entry_Test_Type', 
        'Entry_Test_Tier', 'Entry_Test_Required', 'Entry_Test_Passing_Criteria'
    ]
    for col in categorical_cols:
        df_clean[col] = df_clean[col].astype('category')
        
    # Pipeline execution audit report
    print("Production Data Cleaning Pipeline executed successfully!")
    print(f"Remaining Structural Missing Values: {df_clean[uni_data.columns].isnull().sum().sum()}")
    print(f"Imputation Logs Generated: {len(impute_target_cols)} flag columns created.")
    print(f"Final Dataframe Shape: {df_clean.shape[0]} rows, {df_clean.shape[1]} columns")
    
    return df_clean
```

```python
def execution_analysis_pipeline(c_uni_data):
    """
    Executes the comprehensive 10-question EDA notebook analysis.
    Directly displays all cleaned and corrected summary outputs.
    """
    # Q1: Complete HEC category ranking
    print("=== Q1: HEC Category Ranking by Avg CGPA ===")
    q1_output = (
        c_uni_data.groupby('HEC_Category', observed=False)['Avg_CGPA']
        .mean()
        .sort_values(ascending=False)
    )
    display(q1_output)
    print("\n" + "="*50 + "\n")
    
    # Q2: Number of universities per Province
    print("=== Q2: Number of Universities by Province ===")
    q2_output = (
        c_uni_data.groupby('Province', observed=False)['UNI_Name']
        .nunique()
        .sort_values(ascending=False)
    )
    display(q2_output)
    print("\n" + "="*50 + "\n")
    
    # Q3: Research output by broad department category
    print("=== Q3: Total Research Papers by Department Category ===")
    q3_output = (
        c_uni_data.groupby('Department_Category', observed=False)['Research_Papers_Published']
        .sum()
        .sort_values(ascending=False)
    )
    display(q3_output)
    print("\n" + "="*50 + "\n")
    
    # Q4: Median semester fee by sector Type
    print("=== Q4: Median Fee Per Semester by University Type ===")
    q4_output = (
        c_uni_data.groupby('Type', observed=False)['Fee_Per_Semester_PKR']
        .median()
    )
    display(q4_output)
    print("\n" + "="*50 + "\n")
    
    # Q5: Top 5 cities by average Employment Ratio (with >= 3 universities)
    print("=== Q5: Top 5 Cities by Average Employment Ratio (Min 3 Unis) ===")
    city_counts = c_uni_data.groupby('City', observed=False)['UNI_Name'].nunique()
    eligible_cities = city_counts[city_counts >= 3].index
    q5_output = (
        c_uni_data[c_uni_data['City'].isin(eligible_cities)]
        .groupby('City', observed=False)['Employment_Ratio_Pct']
        .mean()
        .nlargest(5)
    )
    display(q5_output)
    print("\n" + "="*50 + "\n")
    
    # Q6: Dropout rates - Overall and Sector Breakdown
    print("=== Q6 (Part 1): Overall Average Dropout Rate by Province ===")
    q6_overall = (
        c_uni_data.groupby('Province', observed=True)['Dropout_Rate_Pct']
        .mean()
        .sort_values(ascending=False)
    )
    display(q6_overall)
    print("\n--- Q6 (Part 2): Breakdown by Province and Type ---")
    q6_breakdown = (
        c_uni_data.groupby(['Province', 'Type'], observed=True)['Dropout_Rate_Pct']
        .mean()
        .sort_values(ascending=False)
    )
    display(q6_breakdown)
    print("\n" + "="*50 + "\n")
    
    # Q7: Academic and market metrics across entry test difficulty tiers
    print("=== Q7: Metrics by Entry Test Tier ===")
    q7_output = (
        c_uni_data.groupby('Entry_Test_Tier', observed=False)[['Avg_CGPA', 'Employment_Ratio_Pct']]
        .mean()
    )
    display(q7_output)
    print("\n*Note: Clear positive correlation observed. Higher entrance difficulty links directly to stronger CGPAs and employment metrics.*")
    print("\n" + "="*50 + "\n")
    
    # Q8: High enrollment aggregate summary (>50k students)
    print("=== Q8: Summary Table (Enrollments > 50,000) ===")
    summary_table = (
        c_uni_data.groupby(['Province', 'Department_Category'], observed=True)
        .agg(
            Total_Enrollment=('Total_Enrollment', 'sum'),
            Avg_Student_Faculty_Ratio=('Student_Faculty_Ratio', 'mean'),
            Number_of_Departments=('Department_Name', 'nunique') # Unique check fixed
        )
    )
    q8_output = summary_table[summary_table['Total_Enrollment'] > 50000]
    display(q8_output)
    print("\n" + "="*50 + "\n")
    
    # Q9: Count unique universities exceeding provincial female share baselines
    print("=== Q9: Universities Exceeding Provincial Female Share Baseline ===")
    uni_level = (
        c_uni_data.groupby(['Province', 'UNI_Name'], observed=True)
        .agg(Total_Female=('Female_Enrollment', 'sum'), Total_Students=('Total_Enrollment', 'sum'))
        .reset_index()
    )
    uni_level['Female_Share'] = uni_level['Total_Female'] / uni_level['Total_Students']
    uni_level['Provincial_Avg'] = uni_level.groupby('Province', observed=True)['Female_Share'].transform('mean')
    
    q9_output = (
        uni_level[uni_level['Female_Share'] > uni_level['Provincial_Avg']]
        .groupby('Province', observed=True)['UNI_Name']
        .nunique()
        .sort_values(ascending=False)
    )
    display(q9_output)
    print("\n" + "="*50 + "\n")
    
    # Q10: Top research category per year
    print("=== Q10: Top Department Category for Research Output Yearly ===")
    q10_output = (
        c_uni_data.groupby(['Year', 'Department_Category'], observed=True)['Research_Papers_Published']
        .mean()
        .reset_index()
        .sort_values('Research_Papers_Published', ascending=False)
        .drop_duplicates(subset=['Year'], keep='first')
        .sort_values('Year')
        .reset_index(drop=True)
    )
    display(q10_output)
```

```python
def run_feature_engineering_pipeline(df_path, output_path):
    """
    Consolidates all 10 feature engineering exercises into a single pipeline,
    executes them, and saves the enhanced dataset to a CSV file.
    """
    # Load the cleaned dataset
    df = pd.read_csv(df_path)
    
    print("--- Starting Feature Engineering Pipeline ---")

    # 1 & 2. Female Share Percentage & Binning
    df['Female_Share_Pct'] = (df['Female_Enrollment'] / df['Total_Enrollment'].replace(0, np.nan)).fillna(0) * 100
    df['Gender_Balance_Category'] = pd.cut(
        df['Female_Share_Pct'], 
        bins=[-1, 30, 60, 101], 
        labels=['Male Dominated', 'Balanced', 'Female Dominated']
    )
    print("Calculated Female Share and Gender Categories.")

    # 3. Annual Fee & Fee Bands
    df['Annual_Fee_PKR'] = df['Fee_Per_Semester_PKR'] * 2
    df['Fee_Band'] = pd.qcut(
        df['Annual_Fee_PKR'], 
        q=[0, 0.33, 0.66, 1.0], 
        labels=['Low', 'Mid', 'High']
    )
    print("Calculated Annual Fee and Percentile Fee Bands.")

    # 4. Research Productivity Per Faculty
    df['Research_Per_Faculty'] = (
        df['Research_Papers_Published'] / df['Faculty_Count'].replace(0, np.nan)
    ).fillna(0)
    print("Calculated Research Per Faculty.")

    # 5. Campus Density & Overcrowding Flag
    df['Campus_Density'] = (
        df['Total_Enrollment'] / df['No_of_Buildings'].replace(0, np.nan)
    ).fillna(0)
    national_median_density = df['Campus_Density'].median()
    df['Is_Overcrowded'] = (df['Campus_Density'] > (1.5 * national_median_density)).astype(int)
    print(f"Flagged Overcrowded Campuses (National Median Density Baseline: {national_median_density:.2f}).")

    # 6. Postgraduate Ratio & Research Focus Flag
    df['PG_Ratio'] = (
        ((df['MS_MPhil_Students'] + df['PhD_Students']) / df['Total_Enrollment'].replace(0, np.nan)) * 100
    ).fillna(0)
    median_research_per_faculty = df['Research_Per_Faculty'].median()
    df['Is_Research_Focused'] = (
        (df['PG_Ratio'] > 30) & (df['Research_Per_Faculty'] > median_research_per_faculty)
    ).astype(int)
    print("Calculated PG Ratio and Research Focus Flags.")

    # 7. Selectivity Score
    tier_mapping = {'Difficult': 3, 'Medium': 2, 'Easy': 1, 'None': 0}
    encoded_tier = df['Entry_Test_Tier'].astype(str).map(tier_mapping).fillna(0)
    cgpa_min, cgpa_max = df['Avg_CGPA'].min(), df['Avg_CGPA'].max()
    normalized_cgpa = (df['Avg_CGPA'] - cgpa_min) / (cgpa_max - cgpa_min) if (cgpa_max - cgpa_min) != 0 else 0
    df['Selectivity_Score'] = (0.5 * encoded_tier) + (0.5 * normalized_cgpa)
    print("Calculated Selectivity Score.")

    # 8. University Score Composite Ranking (Aggregated to University Level)
    metrics = ['Employment_Ratio_Pct', 'Research_Per_Faculty', 'Avg_CGPA', 'Industry_Tie_Ups']
    uni_means = df.groupby('UNI_ID')[metrics].transform('mean')
    
    norm_metrics = {}
    for col in metrics:
        c_min, c_max = uni_means[col].min(), uni_means[col].max()
        norm_metrics[col] = (uni_means[col] - c_min) / (c_max - c_min) if (c_max - c_min) != 0 else 0
        
    df['University_Score'] = (
        (0.35 * norm_metrics['Employment_Ratio_Pct']) +
        (0.25 * norm_metrics['Research_Per_Faculty']) +
        (0.25 * norm_metrics['Avg_CGPA']) +
        (0.15 * norm_metrics['Industry_Tie_Ups'])
    )
    print("Calculated Composite University Score.")

    # 9. Fee Value Score
    df['Fee_Value_Score'] = (df['Employment_Ratio_Pct'] / df['Annual_Fee_PKR']) * 100000
    print("Calculated Fee Value Score.")

    # 10. Peer CGPA Gap (Regional Tier Benchmarking)
    provincial_tier_mean = df.groupby(['Province', 'HEC_Category'], observed=False)['Avg_CGPA'].transform('mean')
    df['Peer_CGPA_Gap'] = df['Avg_CGPA'] - provincial_tier_mean
    print("Calculated Peer CGPA Gap Benchmark.")

    # 11. Risk Index Composite
    norm_dropout = (df['Dropout_Rate_Pct'] - df['Dropout_Rate_Pct'].min()) / (df['Dropout_Rate_Pct'].max() - df['Dropout_Rate_Pct'].min())
    norm_sfr = (df['Student_Faculty_Ratio'] - df['Student_Faculty_Ratio'].min()) / (df['Student_Faculty_Ratio'].max() - df['Student_Faculty_Ratio'].min())
    inverted_female = 1 - (df['Female_Share_Pct'] - df['Female_Share_Pct'].min()) / (df['Female_Share_Pct'].max() - df['Female_Share_Pct'].min())
    df['Risk_Index'] = (norm_dropout + norm_sfr + inverted_female) / 3
    print("Calculated Institutional Risk Index.")

    # Save to disk
    df.to_csv(output_path, index=False)
    print(f"Success! Engineered dataset saved with shape {df.shape} to: '{output_path}'")
    return df
```

```python
import pandas as pd
import plotly.express as px


# =====================================================================
# FUNCTION 1: DEMOGRAPHICS & SHARE
# =====================================================================
def plot_demographics_and_share(data: pd.DataFrame) -> None:
    """Aggregates and visualizes university distribution across provinces

    and splits them by public/private sector type.
    """
    # Group by province and count unique universities
    total_uni_pro = (
        data.groupby("Province", observed=False)["UNI_Name"]
        .nunique()
        .sort_values(ascending=False)
    )

    # Generate a vertical bar chart of university counts
    fig_bar = px.bar(
        total_uni_pro,
        x=total_uni_pro.index,
        y=total_uni_pro.values,
        color=total_uni_pro.index,
        text_auto=True,
        title="<b>Number of Universities Per Province</b>",
        labels={
            "x": "Province",
            "y": "Number of Universities",
        },
        color_discrete_sequence=px.colors.qualitative.Plotly,
    )

    # Style the bar chart layout for a light theme
    fig_bar.update_layout(
        title_font=dict(size=22, family="Arial", color="#2c3e50"),
        title_x=0.5,
        showlegend=False,
        plot_bgcolor="rgba(0,0,0,0)",
        paper_bgcolor="rgba(0,0,0,0)",
        margin=dict(l=50, r=50, t=80, b=50),
    )
    fig_bar.update_xaxes(
        showline=True,
        linewidth=2,
        linecolor="#bdc3c7",
        gridcolor="rgba(0,0,0,0)",
        tickfont=dict(size=12, family="Arial", color="#2c3e50"),
    )
    fig_bar.update_yaxes(
        showgrid=True,
        gridcolor="#ecf0f1",
        tickfont=dict(size=12, family="Arial", color="#2c3e50"),
    )
    fig_bar.update_traces(
        textposition="outside",
        textfont=dict(size=13, weight="bold", color="#2c3e50"),
        marker_line_color="rgba(0,0,0,0)",
    )
    fig_bar.show()

    # Calculate unique university breakdown by public/private type
    share_type = (
        data.groupby("Type", observed=False)["UNI_Name"]
        .nunique()
        .reset_index(name="University_Count")
    )

    # Generate a pie chart for sector market share
    fig_pie = px.pie(
        share_type,
        names="Type",
        values="University_Count",
        title="<b>Share of Universities by Type (Public vs Private)</b>",
        color_discrete_sequence=px.colors.qualitative.Safe,
    )

    # Style the pie chart layout and position the legend
    fig_pie.update_layout(
        title_font=dict(size=22, family="Arial", color="#2c3e50"),
        title_x=0.5,
        legend=dict(
            orientation="h", yanchor="bottom", y=-0.1, xanchor="center", x=0.5
        ),
    )
    fig_pie.update_traces(
        textinfo="label+percent",
        textposition="inside",
        textfont=dict(size=14, weight="bold", color="#ffffff"),
        marker=dict(line=dict(color="#ffffff", width=2)),
        automargin=True,
    )
    fig_pie.show()


# =====================================================================
# FUNCTION 2: ACADEMICS & FINANCES
# =====================================================================
def plot_academics_and_finances(data: pd.DataFrame) -> None:
    """Visualizes academic performance and financial statistics including

    research volumes, fee spreads, and employability indicators.
    """
    # Calculate total research papers for the top department categories
    research_dept = (
        data.groupby("Department_Category", observed=False)[
            "Research_Papers_Published"
        ]
        .sum()
        .reset_index()
    )
    research_dept = research_dept.sort_values(
        by="Research_Papers_Published", ascending=False
    ).head(10)

    # Create a horizontal bar chart of research volume by department
    fig_research = px.bar(
        research_dept,
        x="Research_Papers_Published",
        y="Department_Category",
        orientation="h",
        color="Department_Category",
        text_auto=True,
        title="<b>Top Department Categories by Research Papers Published</b>",
        labels={
            "Research_Papers_Published": "Total Research Papers",
            "Department_Category": "Department Category",
        },
        color_discrete_sequence=px.colors.qualitative.Plotly,
    )
    fig_research.update_layout(
        title_font=dict(size=22, family="Arial", color="#2c3e50"),
        title_x=0.5,
        showlegend=False,
        plot_bgcolor="rgba(0,0,0,0)",
        paper_bgcolor="rgba(0,0,0,0)",
        margin=dict(l=160, r=60, t=80, b=50),
        yaxis=dict(categoryorder="total ascending"),
    )
    fig_research.update_xaxes(
        showgrid=True, gridcolor="#ecf0f1", tickfont=dict(size=12, family="Arial")
    )
    fig_research.update_yaxes(
        showline=True,
        linewidth=2,
        linecolor="#bdc3c7",
        tickfont=dict(size=12, family="Arial"),
    )
    fig_research.update_traces(
        textposition="outside",
        textfont=dict(size=12, weight="bold", color="#2c3e50"),
        marker_line_color="rgba(0,0,0,0)",
    )
    fig_research.show()

    # Create a box plot comparing semester tuition fees by type
    fig_box = px.box(
        data,
        x="Type",
        y="Fee_Per_Semester_PKR",
        color="Type",
        title="<b>Distribution of Fee Per Semester by University Type</b>",
        labels={
            "Fee_Per_Semester_PKR": "Fee Per Semester (PKR)",
            "Type": "University Type",
        },
        color_discrete_sequence=px.colors.qualitative.Safe,
    )
    fig_box.update_layout(
        title_font=dict(size=22, family="Arial", color="#2c3e50"),
        title_x=0.5,
        showlegend=False,
        plot_bgcolor="rgba(0,0,0,0)",
        paper_bgcolor="rgba(0,0,0,0)",
        margin=dict(l=80, r=50, t=80, b=50),
    )
    fig_box.update_xaxes(
        showline=True,
        linewidth=2,
        linecolor="#bdc3c7",
        tickfont=dict(size=14, family="Arial", color="#2c3e50"),
    )
    fig_box.update_yaxes(
        showgrid=True, gridcolor="#ecf0f1", tickfont=dict(size=12, family="Arial")
    )
    fig_box.show()

    # Create a scatter plot matching employment ratios against student grades
    fig_scatter = px.scatter(
        data,
        x="Avg_CGPA",
        y="Employment_Ratio_Pct",
        color="HEC_Category",
        hover_data=["UNI_Name", "City", "Province"],
        title="<b>Employment Ratio vs Average CGPA by HEC Category</b>",
        labels={
            "Avg_CGPA": "Average CGPA",
            "Employment_Ratio_Pct": "Employment Ratio (%)",
            "HEC_Category": "HEC Category",
            "UNI_Name": "University",
            "City": "City",
            "Province": "Province",
        },
        color_discrete_sequence=px.colors.qualitative.Safe,
    )
    fig_scatter.update_layout(
        title_font=dict(size=22, family="Arial", color="#2c3e50"),
        title_x=0.5,
        legend_title_text="HEC Category",
        plot_bgcolor="rgba(0,0,0,0)",
        paper_bgcolor="rgba(0,0,0,0)",
        margin=dict(l=80, r=50, t=80, b=50),
    )
    fig_scatter.update_xaxes(
        showline=True,
        linewidth=2,
        linecolor="#bdc3c7",
        showgrid=True,
        gridcolor="#ecf0f1",
    )
    fig_scatter.update_yaxes(
        showline=True,
        linewidth=2,
        linecolor="#bdc3c7",
        showgrid=True,
        gridcolor="#ecf0f1",
    )
    fig_scatter.update_traces(
        marker=dict(
            size=8, opacity=0.75, line=dict(width=0.5, color="#ffffff")
        )
    )
    fig_scatter.show()

    # Create an overlaid histogram showing faculty research distributions
    fig_hist = px.histogram(
        data,
        x="Research_Per_Faculty",
        color="HEC_Category",
        nbins=30,
        barmode="overlay",
        opacity=0.65,
        title="<b>Distribution of Research Output Per Faculty by HEC Category</b>",
        labels={
            "Research_Per_Faculty": "Research Papers Per Faculty",
            "count": "Frequency (Count)",
            "HEC_Category": "HEC Category",
        },
        color_discrete_sequence=px.colors.qualitative.Safe,
    )
    fig_hist.update_layout(
        title_font=dict(size=22, family="Arial", color="#2c3e50"),
        title_x=0.5,
        legend_title_text="HEC Category",
        plot_bgcolor="rgba(0,0,0,0)",
        paper_bgcolor="rgba(0,0,0,0)",
        margin=dict(l=80, r=50, t=80, b=50),
    )
    fig_hist.update_xaxes(
        showline=True,
        linewidth=2,
        linecolor="#bdc3c7",
        showgrid=True,
        gridcolor="#ecf0f1",
    )
    fig_hist.update_yaxes(
        showline=True,
        linewidth=2,
        linecolor="#bdc3c7",
        showgrid=True,
        gridcolor="#ecf0f1",
    )
    fig_hist.show()


# =====================================================================
# FUNCTION 3: ADVANCED STRUCTURAL TRENDS
# =====================================================================
def plot_advanced_trends(data: pd.DataFrame) -> None:
    """Generates complex multivariable plots including bubble charts,

    faceted grid layouts, and longitudinal trend lines.
    """
    # Sort provinces by overall average dropout rates
    prov_order = (
        data.groupby("Province", observed=False)["Dropout_Rate_Pct"]
        .mean()
        .sort_values(ascending=False)
        .index.tolist()
    )

    # Group dropout records by province and sector type
    drop_uni = (
        data.groupby(["Province", "Type"], observed=False)["Dropout_Rate_Pct"]
        .mean()
        .reset_index()
    )

    # Create a grouped bar chart comparing dropout rates side by side
    fig_dropout = px.bar(
        drop_uni,
        x="Province",
        y="Dropout_Rate_Pct",
        color="Type",
        barmode="group",
        text_auto=".1f",
        title="<b>Average Dropout Rate by Province and University Type</b>",
        labels={
            "Dropout_Rate_Pct": "Average Dropout Rate (%)",
            "Province": "Province",
            "Type": "University Type",
        },
        color_discrete_sequence=px.colors.qualitative.Safe,
    )
    fig_dropout.update_layout(
        title_font=dict(size=22, family="Arial", color="#2c3e50"),
        title_x=0.5,
        legend=dict(
            orientation="h", yanchor="bottom", y=-0.2, xanchor="center", x=0.5
        ),
        plot_bgcolor="rgba(0,0,0,0)",
        paper_bgcolor="rgba(0,0,0,0)",
        margin=dict(l=70, r=50, t=80, b=70),
    )
    fig_dropout.update_xaxes(
        showline=True,
        linewidth=2,
        linecolor="#bdc3c7",
        gridcolor="rgba(0,0,0,0)",
        categoryorder="array",
        categoryarray=prov_order,
        tickfont=dict(size=12, family="Arial"),
    )
    fig_dropout.update_yaxes(
        showgrid=True, gridcolor="#ecf0f1", tickfont=dict(size=12, family="Arial")
    )
    fig_dropout.update_traces(
        textposition="outside",
        textfont=dict(size=11, weight="bold"),
        marker_line_color="rgba(0,0,0,0)",
    )
    fig_dropout.show()

    # Create a bubble chart matching fee levels to employment outcomes
    fig_bubble = px.scatter(
        data,
        x="Fee_Per_Semester_PKR",
        y="Employment_Ratio_Pct",
        color="Province",
        size="Total_Enrollment",
        hover_data=["UNI_Name"],
        title="<b>Employment Ratio vs Fee Per Semester by Province</b>",
        labels={
            "Fee_Per_Semester_PKR": "Fee Per Semester (PKR)",
            "Employment_Ratio_Pct": "Employment Ratio (%)",
            "Province": "Province",
            "Total_Enrollment": "Total Enrollment",
            "UNI_Name": "University",
        },
        color_discrete_sequence=px.colors.qualitative.Safe,
        size_max=30,
    )
    fig_bubble.update_layout(
        title_font=dict(size=22, family="Arial", color="#2c3e50"),
        title_x=0.5,
        legend_title_text="Province",
        plot_bgcolor="rgba(0,0,0,0)",
        paper_bgcolor="rgba(0,0,0,0)",
        margin=dict(l=80, r=50, t=80, b=50),
    )
    fig_bubble.update_xaxes(
        showline=True,
        linewidth=2,
        linecolor="#bdc3c7",
        showgrid=True,
        gridcolor="#ecf0f1",
    )
    fig_bubble.update_yaxes(
        showline=True,
        linewidth=2,
        linecolor="#bdc3c7",
        showgrid=True,
        gridcolor="#ecf0f1",
    )
    fig_bubble.update_traces(
        marker=dict(opacity=0.7, line=dict(width=0.8, color="#ffffff"))
    )
    fig_bubble.show()

    # Group average engineered score metrics across regions
    facet_data = (
        data.groupby(["Province", "HEC_Category"], observed=False)[
            "University_Score"
        ]
        .mean()
        .reset_index()
    )

    # Create a wrapped multi-row subplot grid grouped by province
    fig_facet = px.bar(
        facet_data,
        x="HEC_Category",
        y="University_Score",
        color="HEC_Category",
        facet_col="Province",
        facet_col_wrap=3,
        facet_row_spacing=0.10,
        facet_col_spacing=0.06,
        text_auto=".1f",
        labels={"HEC_Category": "HEC Category", "Province": "Province"},
        color_discrete_sequence=px.colors.qualitative.Safe,
    )
    fig_facet.update_layout(
        title=dict(
            text="<b>Average University Score by HEC Category across Provinces</b>",
            x=0.5,
            y=0.96,
            xanchor="center",
            yanchor="top",
            font=dict(size=22, family="Arial", color="#2c3e50"),
        ),
        showlegend=False,
        plot_bgcolor="rgba(0,0,0,0)",
        paper_bgcolor="rgba(0,0,0,0)",
        margin=dict(l=80, r=40, t=130, b=60),
    )
    fig_facet.update_yaxes(title_text="")
    fig_facet.add_annotation(
        text="<b>Avg University Score</b>",
        textangle=-90,
        xref="paper",
        yref="paper",
        x=-0.06,
        y=0.5,
        showarrow=False,
        font=dict(size=14, family="Arial", color="#2c3e50"),
    )
    fig_facet.for_each_annotation(
        lambda a: a.update(
            text=f"<b>{a.text.split('=')[-1]}</b>",
            font=dict(size=13, color="#2c3e50"),
        )
    )
    fig_facet.update_xaxes(
        showline=True,
        linewidth=1.5,
        linecolor="#bdc3c7",
        tickfont=dict(size=11, family="Arial"),
    )
    fig_facet.update_yaxes(
        showgrid=True, gridcolor="#ecf0f1", tickfont=dict(size=10, family="Arial")
    )
    fig_facet.update_traces(
        textposition="outside",
        textfont=dict(size=10, weight="bold"),
        marker_line_color="rgba(0,0,0,0)",
    )
    fig_facet.show()

    # Track longitudinal research metrics broken down by department type
    line_data = (
        data.groupby(["Year", "Department_Category"], observed=False)[
            "Research_Papers_Published"
        ]
        .mean()
        .reset_index()
    )

    # Generate a multi-line trend graph with key metrics marked
    fig_line = px.line(
        line_data,
        x="Year",
        y="Research_Papers_Published",
        color="Department_Category",
        markers=True,
        title="<b>Average Research Papers Published Over Time by Department Category</b>",
        labels={
            "Year": "Year",
            "Research_Papers_Published": "Avg Research Papers Published",
            "Department_Category": "Department Category",
        },
        color_discrete_sequence=px.colors.qualitative.Safe,
    )
    fig_line.update_layout(
        title_font=dict(size=22, family="Arial", color="#2c3e50"),
        title_x=0.5,
        legend=dict(
            orientation="h", yanchor="top", y=-0.3, xanchor="center", x=0.5
        ),
        plot_bgcolor="rgba(0,0,0,0)",
        paper_bgcolor="rgba(0,0,0,0)",
        margin=dict(l=80, r=40, t=80, b=100),
    )
    fig_line.update_xaxes(
        showline=True,
        linewidth=2,
        linecolor="#bdc3c7",
        type="category",
        tickfont=dict(size=12, family="Arial"),
    )
    fig_line.update_yaxes(
        showgrid=True,
        gridcolor="#ecf0f1",
        showline=True,
        linewidth=2,
        linecolor="#bdc3c7",
        tickfont=dict(size=12, family="Arial"),
    )
    fig_line.update_traces(
        line=dict(width=3.5),
        marker=dict(size=9, line=dict(width=1, color="#ffffff")),
    )
    fig_line.show()
```
